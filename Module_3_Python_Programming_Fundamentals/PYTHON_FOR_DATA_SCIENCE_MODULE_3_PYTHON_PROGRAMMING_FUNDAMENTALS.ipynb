```python
# CONDITONS AND BRANCHING
# Conditions and branching allow your program to make decisions and run different code depending on whether something is True or False.

# EQUALITY OPERATOR
# == is the equality operator.It checks whether two values are equal.
# It returns either:
# True → if they are equal
# False → if they are not equal
a=6
a==7
```

```python
# GREATER THAN EQUALITY OPERATOR
# Consider the following equality comparison operator.If the value of the left operand, in this case, the variable “i” is greater than
# the value of the right operand, in this case, 5, the condition becomes true, or else we get a false
I = 6
I > 5
```

```python
# GREATER THAN OR EUAL TO  OPERATOR
# If we modify the operator as follows, if the left operand “i” is greater than or equal to the value of the right operand, in this case, 5, then the condition becomes true.
I = 6
I >= 5
```

```python
# LESS THAN EQUALITY OPERATOR
# We can change the inequality, if the value of the left operand, in this case, “i” is less than the value of right operand, in this case, 6, then condition becomes true.
```

```python
# NOT EQUAL TO
# In Python, “not equal to” is written as:!=
# If we set “I” equal to 2, the operator is true as 2 is not equal to 6.
I = 2
I != 6

```

```python
# DIFFERENCE BETWEEN = AND ==
# ✅ = → Assignment Operator: It is used to assign (store) a value to a variable.
# Example: x = 5

# ✅ == → Comparison Operator (Equality Test):It is used to compare two values to check if they are equal.
# Example: x == 5
# 👉 Meaning: “Is x equal to 5?”
# This returns either:

# True (if x is 5)

# False (if x is not 5)
```

```python
# Very short summary:

# = → puts a value into a variable

# == → checks if two values are equal
```

```python
# What is a condition?
# A condition is a statement that can be True or False.
```

```python
# BRANCHING:Branching means making different decisions in your program depending on a condition.
# In Python, branching is done using:

# 1. if

# 2.elif

# 3.else

# Your program “branches” into different paths based on whether a condition is True or False.
```

```python
# IF STATEMENT: An if statement tells Python to do something only when a condition is True.
# ✅ Basic Structure
# if condition:
    # do this if the condition is True
# 👉 If the condition is False, Python simply skips this part.
```

```python
Age = 17
if (Age > 18):
    print("You can enter")
print("move on")
```

```python
Age = 19
if (Age > 18):
    print("You can enter")
print("move on")
```

```python
# THE ELSE STATEMENT: The else statement runs only when the if condition is False.
# Think of it like this:

# If = this is true → do this
# Else = if it is NOT true → do that
```

```python
# ✅ Basic Structure
# if condition:
    # runs when condition is True
# else:
    # runs when condition is False
```

```python
age = 16

if age > 18:
    print("You can enter")
else:
    print("You cannot enter")
```

```python
# THE ELIF STATEMENT: elif means “else if.” It is used when you want to check more than one condition.

# ✅ Basic Structure
# if condition1:
    # runs if condition1 is True
# elif condition2:
    # runs if condition1 is False AND condition2 is True
# else:
    # runs if all previous conditions are False


# 👉 Only ONE block will run.
```

```python
age = 18

if age > 18:
    print("You are an adult")

elif age == 18:
    print("You are just 18")

else:
    print("You are a minor")
```

```python
age = 18

if age > 18:
    print("You can see the ACDC concert")

elif age == 18:
    print("You can see the pink floys concert")

else:
    print("You can see the Meat Loaf concert")
```

```python
# LOGIC OPERATORS: Logical operators are used to combine or change conditions in an if statement.

# There are three main logical operators in Python:
# and:both conditions must be True
# or:at least one condition must be True
# not:makes True become False, and False become True
```

```python
# ✅ 1) AND OPERATOR:Both sides must be True.

age = 20
have_ticket = True

if age > 18 and have_ticket:
    print("You can enter the concert")


else:
    print("You cannot enter")


# 👉 This only works if:

# age is greater than 18 AND

# you have a ticket

# If either one is False → you cannot enter.
```

```python
album_year = 1990
if (album_year < 1980) and (album_year > 1988):
    print("The album was made in the 70's or 90's")
else:
    print("The Album was made in the 1980's")
```

```python
# ✅ 2) OR:Only one side needs to be True.

rain = True
umbrella = False

if rain or umbrella:
    print("Be careful outside")


# 👉 This prints because rain is True, even though umbrella is False.
```

```python
album_year = 1990
if (album_year < 1980) or (album_year > 1988):
    print("The album was made in the 70's or 90's")
else:
    print("The Album was made in the 1980's")
```

```python
# ✅ 3) NOT OPERATOR:It flips (reverses) the condition.

raining = False

if not raining:
    print("You can go outside")


# Since raining is False,
# 👉 not raining becomes True → so it prints.
```
