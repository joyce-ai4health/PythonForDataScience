```python
# 1. Find the maximum value in a list
numbers = [2, 4, 7, 4, 23, 5, 1, 4, 8, 9]
maximum = numbers[0]
for num in numbers:
    if num > maximum:
        maximum = num

print("Maximum value:", maximum)
```

```python
# 2. Calculate the average value of a list
numbers = [4, 7, 1, 5, 11, 53, 12, 46, 84, 23]
total = 0
for num in numbers:
    total += num

average = total / len(numbers)
print("Average value:", average)
```

```python
# 3. Print a list in reverse order
numbers = [2, 6, 7, 45, 23, 53, 14, 45, 89, 5]
for i in range(len(numbers) - 1, -1, -1):
    print(numbers[i])
```

```python
# 4. Compare two lists element by element
list1 = [1, 3, 5]
list2 = [2, 4, 6]
if len(list1) != len(list2):
    print(False)
else:
    result = True
    for i in range(len(list1)):
        if list1[i] >= list2[i]:
            result = False
            break
    print(result)
```

```python
# 5. Swap two elements in a list using indexes
numbers = [10, 20, 30, 40, 50]
index1 = 1
index2 = 3
numbers[index1], numbers[index2] = numbers[index2], numbers[index1]

print("After swapping:", numbers)
```

```python
# 6. Combine two lists into one
list1 = [1, 2, 3]
list2 = [4, 5, 6]

new_list = list1 + list2
print("Combined list:", new_list)
```

```python
# 7. Find the last index of a value in a list
numbers = [74, 85, 102, 99, 101, 85, 56]
value = 85
last_index = -1

for i in range(len(numbers)):
    if numbers[i] == value:
        last_index = i

print("Last index:", last_index)
```

```python
# 8. Print the range of values in a list
numbers = [36, 12, 25, 19, 46, 31, 22]

maximum = max(numbers)
minimum = min(numbers)
range_value = (maximum - minimum) + 1

print("Range:", range_value)
```

```python
# 9. Count values within a given range (inclusive)
numbers = [14, 1, 22, 17, 36, 7, -43, 5]
min_value = 4
max_value = 17
count = 0

for num in numbers:
    if min_value <= num <= max_value:
        count += 1

print("Count:", count)
```

```python
# 10. Check if a list is sorted (nondecreasing order)
def is_sorted(lst):
    for i in range(len(lst) - 1):
        if lst[i] > lst[i + 1]:
            return False
    return True

list1 = [16.1, 12.3, 22.2, 14.4]
list2 = [1.5, 4.3, 7.0, 19.5, 25.1, 46.2]

print(is_sorted(list1))  # False
print(is_sorted(list2))  # True
```
